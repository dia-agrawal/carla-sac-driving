```python
import carla
from carla import Transform, Location, Rotation
import random

client = carla.Client('localhost', 2000)
client.set_timeout(5.0)
world = client.get_world()

transform = Transform(Location(x=50, y=25, z=40), Rotation(yaw=180))

bp_lib = world.get_blueprint_library()

# Option A: exact id (best)
bp = bp_lib.find('vehicle.micro.microlino')

actor = world.try_spawn_actor(bp, transform)  # safer than spawn_actor
print("spawned:", actor)
```

```python
spectator = world.get_spectator()

t = actor.get_transform()
# behind + a little up
behind = t.location - t.get_forward_vector() * 8 + carla.Location(z=3)

spectator.set_transform(carla.Transform(
    behind,
    carla.Rotation(pitch=-10, yaw=t.rotation.yaw)
))
```

```python
vehicle_pos = actor.get_transform()
print("Vehicle position:", vehicle_pos)
```

```python
while True:
    actor.set_autopilot(True)
    t = actor.get_transform()
    # behind + a little up
    behind = t.location - t.get_forward_vector() * 8 + carla.Location(z=3)

    spectator.set_transform(carla.Transform(
        behind,
        carla.Rotation(pitch=-10, yaw=t.rotation.yaw)
    ))
    
```

```python
actor.destroy()
```
